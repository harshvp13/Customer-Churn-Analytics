```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customer = pd.read_csv("Customer Churn.csv")
```

```python
customer.head()
```

```python
customer.info()
```

```python
#fill the missing values in "TotalCharges" with 0 as tenure is 0 and change the data type to float
customer["TotalCharges"] = customer["TotalCharges"].replace(" ","0")
customer['TotalCharges'] = customer["TotalCharges"].astype("float")
```

```python
#check if it has changed the datatype
customer.info()
```

```python
#count the number of missing values in each column
customer.isna().sum()
```

```python
#Statistics
customer.describe()
```

```python
#check if there are duplicates
customer.duplicated(subset="customerID").sum()
```

```python
#Retrun "No" if the customer is not senior citizen else "Yes"
np.where(customer['SeniorCitizen'] == 0,'No','Yes')
customer.head(5)
```

```python
#visualize the number of customers churned and not churned
ax = sns.countplot(x = 'Churn' , data = customer)
ax.bar_label(ax.containers[0])
plt.title("Count of Customers by Churn")
plt.show()
```

```python
#visualize the percentage of customers churned and not churned
gb = customer.groupby("Churn").agg({'Churn' : 'count'})
plt.pie(gb['Churn'], labels= gb.index, autopct='%1.2f%%')
plt.title("Percentage of Customers Churned")
plt.figure(figsize=(3,4))
plt.show()
```

26.54% Customers have Churned out

```python
#Visualize Churn by Gender
sns.countplot(data = customer, x = 'gender' , hue = 'Churn')
plt.title("Churn by Gender")
plt.show()
```

Male and Female have almost done same amount churn

```python
#Percentage of non- senior citizen churned and not
total_counts = customer.groupby("SeniorCitizen")['Churn'].value_counts(normalize=True).unstack() * 100
total_counts
```

```python
#visulizr the customers who have churned out based on tenure
sns.histplot(data = customer, x = "tenure" , bins =72, hue='Churn')
plt.figure(figsize=(9,4))
plt.show()
```

People who have used services over long time have stayed where else the ones who have churned out have left at early stages(i.e 1-2 months)

```python
#Visualize count of customers by contract
ax = sns.countplot(data=customer,x='Contract')
ax.bar_label(ax.containers[0])
plt.title("Count of Customers by Contract")
plt.show()
```

```python
#Visualize customers by contract and see who churns and not
ax = sns.countplot(data=customer,x='Contract',hue = 'Churn')
ax.bar_label(ax.containers[0])
plt.title("Customers by Contract who Churn")
plt.show()
```

People who have month-to-month contract are likely to churn
